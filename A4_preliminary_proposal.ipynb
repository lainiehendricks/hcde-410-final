# Final Project Preliminary Proposal

## Motivation & problem

For my final project, I want to study:

**How does the gemstone mix of high-end jewelry auctions differ across major global markets, and which cities appear to specialize in certain gemstones?**

I am interested in fine jewelry and want to compare major auction markets in Hong Kong, Geneva, New York, London, Paris, and Dubai.

I want to see which gemstones are most common in each city and which categories make up the most auction value. I also want to compare transaction volume and transaction value to see if they show the same patterns.

## Data

I plan to use the **Global Spatiotemporal Distribution Dataset of Jewelry Auction from Sotheby’s and Christie’s (2014–2023)**.

Dataset:  
https://doi.org/10.3974/geodb.2025.12.07.V1

The dataset covers six cities and seven categories: white diamonds, colored diamonds, rubies, sapphires, emeralds, jadeite, and pearls. It includes transaction volume, transaction value, and average price across 42,445 sold lots.

This dataset fits my question because I can compare the same jewelry categories across all six cities.

One limitation is that the data only come from Sotheby’s and Christie’s. This means the results represent the high-end auction market, not all jewelry buyers. The data also reflect what the auction houses choose to sell, not just what buyers prefer.

The dataset is publicly available for use with citation, but redistribution requires written permission from the repository.

## Unknowns & dependencies

My main challenge will be deciding how to measure specialization. I plan to compare each gemstone category’s share of transaction volume and transaction value within each city.

I also need to check how detailed the year-by-year data are. If possible, I may also look at how these patterns changed from 2014–2023.

The dataset is already publicly available, so I do not expect data access to be a major problem.
