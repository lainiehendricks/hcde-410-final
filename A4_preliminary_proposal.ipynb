## Motivation & problem

For my final project, I want to study:

**How does the gemstone mix of high-end jewelry auctions differ across major global markets, and which cities appear to specialize in certain gemstones?**

I am interested in fine jewelry and want to compare major auction markets in Hong Kong, Geneva, New York, London, Paris, and Dubai.

I want to see which gemstones are most common in each city and which categories make up the most auction value. I also want to compare transaction volume and transaction value to see if they show the same patterns.

## Data

I plan to use the **Global Spatiotemporal Distribution Dataset of Jewelry Auction from Sotheby’s and Christie’s (2014–2023)**.

Dataset:  
https://doi.org/10.3974/geodb.2025.12.07.V1

The dataset covers six cities and seven categories: white diamonds, colored diamonds, rubies, sapphires, emeralds, jadeite, and pearls. It includes transaction volume, transaction value, and average price across 42,445 sold lots.

This dataset fits my question because I can compare the same jewelry categories across all six cities.

One limitation is that the data only come from Sotheby’s and Christie’s. This means the results represent the high-end auction market, not all jewelry buyers. The data also reflect what the auction houses choose to sell, not just what buyers prefer.

The dataset is publicly available for use with citation, but redistribution requires written permission from the repository.

## Overview & research questions / hypotheses

My project asks:

**How does the gemstone mix of high-end jewelry auctions differ across major global markets, and which cities appear to specialize in certain gemstones?**

I will compare Hong Kong, Geneva, New York, London, Paris, and Dubai using Sotheby’s and Christie’s auction data from 2014–2023.

My main research questions are:

1. Which gemstone categories make up the largest share of transaction volume in each city?
2. Which categories make up the largest share of transaction value in each city?
3. Do transaction volume and transaction value show different patterns of specialization?
4. If the data allow, how did these patterns change from 2014–2023?

I expect some markets to be more strongly associated with certain gemstones. For example, I expect jadeite to be especially concentrated in Hong Kong, while colored diamonds may account for a larger share of value in Geneva.

## Background / related work

Previous research shows that high-end jewelry auction markets are global but still have regional differences. Wang and Chen’s work on the dataset found that New York, Geneva, and Hong Kong form the main centers of the global jewelry auction market. They also found a strong concentration of jadeite in Hong Kong and a value premium for colored diamonds in Geneva.

Industry research also describes Hong Kong as an important market for jadeite and Asian buyers, while Geneva is known for major high-value jewelry sales. At the same time, auction houses increasingly attract international buyers, which means auction location does not necessarily equal buyer location or culture.

This background shapes my project in two ways. First, it gives me specific patterns to test rather than assuming every market has the same jewelry mix. Second, it means I need to be careful about interpreting differences as “consumer preferences.” My analysis will focus on auction-market composition and specialization rather than making claims about entire regional populations.

## Methodology

I will use descriptive and comparative analysis in Python with pandas and matplotlib.

First, I will inspect and clean the dataset, including checking missing values, category labels, city names, and whether transaction values are stored consistently.

For each city, I will calculate:

- each gemstone category’s share of total transaction volume
- each gemstone category’s share of total transaction value
- average transaction price by gemstone category

I will compare these measures across all six cities to identify which categories are overrepresented or especially valuable in particular markets.

To measure specialization, I will compare each category’s share within a city to its share across the full dataset. A category with a much larger local share than its global share would suggest that the city is relatively specialized in that gemstone.

I will present the results mainly with grouped or stacked bar charts and heatmaps so differences between cities and categories are easy to compare. If the year-level data are detailed enough, I will also use line charts to show how selected patterns changed from 2014–2023.

I will interpret transaction volume and transaction value separately because they measure different things. A category can appear frequently but account for a small share of total value, while another may sell less often but dominate the market by value.
